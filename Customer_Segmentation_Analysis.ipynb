```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel(r"C:\Users\nivet\Downloads\custo_seg_dataset_Standardized_Purchase_History (1).xlsx")
print("File Loaded Successfully")
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(8,6))

corr = df[['Age','Income Level','Coverage Amount','Premium Amount']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.scatterplot(
    data=df.sample(5000, random_state=42),   # sample for speed
    x='Income Level',
    y='Premium Amount',
    hue='Segmentation Group',
    size='Coverage Amount',
    alpha=0.7
)

plt.title("Customer Segmentation Scatter Plot")
plt.xlabel("Income Level")
plt.ylabel("Premium Amount")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x='Segmentation Group'
)

plt.title("Customers by Segmentation Group")
plt.xlabel("Segment")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x='Age',
    bins=15,
    kde=True
)

plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
state_count = df['Geographic Information'].value_counts().reset_index()
state_count.columns = ['State', 'Customers']

# Plot
plt.figure(figsize=(14,8))

sns.barplot(
    data=state_count,
    x='Customers',
    y='State'
)

plt.title("Customers by State")
plt.xlabel("Number of Customers")
plt.ylabel("State")
plt.show()
```
